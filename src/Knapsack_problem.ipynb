# Knapsack problem

Given `n items`, each with distinct `weight` and `price`. 
Maximize the price of the items we take into our backpack without exceeding its carry weight of `M`.

Example: 8 items with:

- 4Kg (19€)
- 2Kg (17€)
- 8Kg (30€)
- 3Kg (13€)
- 7Kg (25€)
- 5Kg (29€)
- 9Kg (23€)
- 6Kg (20€) 

```python
from pulp import *

# data
n = 8
weights = [4, 2, 8, 3, 7, 5, 9, 6]
prices = [9, 17, 30, 13, 25, 29, 23, 20]
carry_weight = 17

# creating the model
model = LpProblem(sense=LpMaximize)

# variables - we create a binary variable which refers whether the piece is picked or not
variables = [LpVariable(name = f"x_{i}" , cat =LpBinary) for i in range(n)]

# inequalities (w1*b1 + w2*b2 +... <=17)
model += lpDot(weights, variables) <=17

# objective function - maximize the price of items
model += lpDot(prices, variables) 

# solve (ignoring debug messages)
status = model.solve(PULP_CBC_CMD(msg=False))

print("price:", model.objective.value())
print("binary relation:", [variables[i].value() for i in range(n)])
print("take:", [variables[i] for i in range(n) if variables[i].value() == 1])
```
