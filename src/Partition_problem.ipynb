# Partition problem

Given `n` items with weights `split` them `into two parts` such that the `difference in their weights` is minimized.

```python
from pulp import *

# data
item_sets = [
    ("Set 1", [70, 60, 50, 33, 33, 33, 11, 7, 3]),
    ("Set 2", [99, 94, 79, 64, 50, 46, 43, 37, 32, 19, 18, 7, 6, 3]),
    ("Set 3", [49, 41, 34, 33, 29, 26, 26, 22, 20, 19]),
    ("Set 4", [442, 252, 252, 252, 252, 252, 252, 252, 127, 127, 127, 127, 127, 106, 106, 106, 106, 85, 84, 46, 37, 37, 12, 12, 12, 10, 10, 10, 10, 10, 10, 9, 9]),
]

# Iterate all sets
for set_name, weights in item_sets:
    n = len(weights) # number of total items

    # minimizing model
    model = LpProblem(sense=LpMinimize)

    # variables - we create a binary variable which refers whether the piece is picked or not
    variables = [LpVariable(name = f"x_{i}" , cat =LpBinary) for i in range(n)]

    # ... and also the difference between two splits
    diff = LpVariable(name="difference", cat=LpInteger , lowBound=0)

    # inequalities
    ## the difference of the parts equals their difference (variable)
    model += diff == lpSum([variables[i] * weights[i] for i in range(n)]) \
            - lpSum([(1 - variables[i]) * weights[i] for i in range(n)])

    # objective function - minimize the difference between the sides
    model += diff
    
    # solve the problem
    status = model.solve(PULP_CBC_CMD(msg=False))

    print(set_name + "\n" + "-" * len(set_name))
    print("Difference:", diff.value())
    bin_0 = []
    sum_0 = 0
    bin_1 = []
    sum_1 = 0
    for i in range(n):
        if int(variables[i].value()) == 0:
            bin_0.append(f"{variables[i]} (w:{weights[i]})")
            sum_0 += weights[i]
        else:
            bin_1.append(f"{variables[i]} (w:{weights[i]})")
            sum_1 += weights[i]
    print(f"Split 0 Total weight: {sum_0}\n\t{bin_0} ")
    print(f"Split 1 Total weight: {sum_1}\n\t{bin_1} ")
    print()
    
```
