# Bipartite max perfect matching problem

Given an undirected graph where nodes represent `people` and `jobs`, and edges have `weights` indicating how well a person is suited for a job.

Determine the `most suitable person - job match`, ensuring each person is assigned to exactly one job in the most optimal way.   

```python
from pulp import *
import string
import random

# define data
n = 25
people = [f"worker_{i+1}" for i in range(n)]

# Alphabetical uppercase letters
alphabetical = string.ascii_uppercase
jobs = [f"job_{alphabetical[i]}" for i in range(n)]

# randomly assigned weights
weights = {(person, job): random.randint(1, 100) for person in people for job in jobs}


# creating the model: maximization problem
model = LpProblem(sense=LpMaximize)

# variables - we create a binary variable which refers to person-job relation. Is related or not.
assigns = LpVariable.dicts("assign", (people, jobs), cat='Binary')

# inequalities

# each person is assigned to one job
for p in people:
    model += lpSum(assigns[p][job] for job in jobs) == 1
    
# each job is assigned to only one person
for j in jobs:
    model += lpSum(assigns[person][j] for person in people) == 1
    

# objective function - maximize the total weight of the matching
model += lpSum(weights[person, job] * assigns[person][job] for person in people for job in jobs)

# Solve the problem
model.solve(PULP_CBC_CMD(msg=False))

# Display results
for person in people:
    for job in jobs:
        if assigns[person][job].varValue == 1:
            print(f"{person} is assigned to {job} with weight {weights[person, job]}")
```
